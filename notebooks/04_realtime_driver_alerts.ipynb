# GTSRB Traffic Sign Recognition - Notebook 04: Real-Time Inference & Driver Alert System

This notebook demonstrates single-frame inference and the **Driver Alert & Safety Advisory Engine**.

### Features:
1. **Single Frame Traffic Sign Prediction**.
2. **Driver Safety Alert Engine**: Categorizes predictions into CRITICAL, WARNING, and INFORMATIONAL tiers.
3. **TTS Speech Prompt Generation**.

```python
import sys
from pathlib import Path
project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

from src.predict import TrafficSignPredictor
from src.driver_alert import DriverAlertSystem

predictor = TrafficSignPredictor(checkpoint_path=str(project_root / "models" / "baseline_best.pth"))
alert_sys = DriverAlertSystem()
print("[INFO] Predictor and DriverAlertSystem ready.")
```

## 1. Single Frame Traffic Sign Classification Demo

```python
sample_img = project_root / "dataset" / "Test" / "14" / "00000.png"  # Stop sign sample if exists
if not sample_img.exists():
    sample_img = list((project_root / "dataset" / "Test").rglob("*.png"))[0]

result = predictor.predict(str(sample_img))
print(f"Predicted Class ID  : {result['class_id']}")
print(f"Predicted Sign Name : {result['class_name']}")
print(f"Confidence Score    : {result['confidence_percentage']}%")
```

## 2. Process Safety Alert & Recommendations

```python
alert_payload = alert_sys.process_detection(result['class_id'], result['confidence_score'])
print("=" * 60)
print(f"{alert_payload['icon']} RISK TIER    : {alert_payload['risk_tier']}")
print(f"SIGN TITLE   : {alert_payload['class_name']}")
print(f"DRIVER ADVICE: {alert_payload['advice']}")
print(f"TTS PROMPT   : {alert_payload['speech_prompt']}")
print("=" * 60)
```
